```python
"""

Attitude Towards Risk / Atitude Frente ao Risco

Author / Autor: imoripaolo

"""

import numpy as np
import matplotlib.pyplot as plt

size = 100
w0, w1 = 0, 100
```

```python
sigma0, sigma1, sigma2 = -0.5,0,0.5

w = np.linspace(w0,w1,size) 
averse, neutral, prone = np.zeros(size), np.zeros(size), np.zeros(size) 
um0, um1, um2 = np.zeros(size), np.zeros(size), np.zeros(size)

for i in range(0, size): 
    prone[i] = ( w[i]**( 1-sigma0 ) ) / ( 1 - sigma0 ) 
    neutral[i] = ( w[i]**( 1-sigma1 ) ) / ( 1 - sigma1 ) 
    averse[i] = ( w[i]**( 1-sigma2 ) ) / ( 1 - sigma2 )

Ew = ( w0 + w1 ) / 2
u_Ew_0 = ( Ew**( 1-sigma0 ) ) / ( 1 - sigma0 ) 
u_Ew_1 = ( Ew**( 1-sigma1 ) ) / ( 1 - sigma1 ) 
u_Ew_2 = ( Ew**( 1-sigma2 ) ) / ( 1 - sigma2 )


E_uw_0 = ( prone[0] + prone[size-1] ) / 2 
E_uw_1 = ( neutral[0] + neutral[size-1] ) / 2 
E_uw_2 = ( averse[0] + averse[size-1] ) / 2

m0 = ( prone[size-1] - prone[0] ) / ( w[size-1] - w[0] ) 
m1 = ( neutral[size-1] - neutral[0] ) / ( w[size-1] - w[0] ) 
m2 = ( averse[size-1] - averse[0] ) / ( w[size-1] - w[0] )

for i in range(0, size): 
    um0[i] = m0w[i] 
    um1[i] = m1w[i] 
    um2[i] = m2*w[i]

```

```python
fig, ax = plt.subplots() 
ax.plot(w,um0,'m') 
ax.plot(w,prone,'b') 
ax.set_ylabel('u(E(w))') 
ax.set_xlabel('w') 
plt.axhline(y=u_Ew_0, color='y', label='U(E(w))') 
plt.axvline(x=Ew, color='y', label = 'E(w)') 
plt.axhline(y=E_uw_0, color='g', label='E(U(w))') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
ax.annotate("P(F,u)< 0", xytext=(26, 150), xy=(40, 230), arrowprops=dict(arrowstyle="->")) 
plt.show()

fig, ax = plt.subplots() 
ax.plot(w,um1,'m') 
ax.plot(w,neutral,'b') 
ax.set_ylabel('u(E(w))') 
ax.set_xlabel('w') 
plt.axhline(y=u_Ew_1, color='y', label='U(E(w))') 
plt.axvline(x=Ew, color='y', label = 'E(w)') 
plt.axhline(y=E_uw_1, color='g', label='E(U(w))') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
ax.annotate("P(F,u)= 0", xytext=(60, 40), xy=(50, 50), arrowprops=dict(arrowstyle="->")) 
plt.show()

fig, ax = plt.subplots() 
ax.plot(w,um2,'m') 
ax.plot(w,averse,'b') 
ax.set_ylabel('u(E(w))') 
ax.set_xlabel('w') 
plt.axhline(y=u_Ew_2, color='y', label='U(E(w))') 
plt.axvline(x=Ew, color='y', label = 'E(w)') 
plt.axhline(y=E_uw_2, color='g', label='E(U(w))') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
ax.annotate("P(F,u)> 0", xytext=(25, 8), xy=(40, 10), arrowprops=dict(arrowstyle="->")) 
plt.show()
```
