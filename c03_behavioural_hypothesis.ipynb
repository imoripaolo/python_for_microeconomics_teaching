```python
""""

Behavioural Hypothesis of the Consumer / Hipótese Comportamental do Consumidor

Author / Autor: imoripaolo

"""""

import numpy as np
import matplotlib.pyplot as plt
```

```python
"""
Auxiliary Functions

----

Funções Auxiliares
"""

def cd(x,y):
    output = (x**0.3)*(y**0.7)
    return output

def expenditure(x,y):
    output = 2*x + 3*y
    return output

x1 = np.linspace(0,50, 50)
x2 = np.linspace(0,50, 50)
X, Y = np.meshgrid(x1, x2)

U = cd(X, Y)
E = budget(X,Y) 



```

```python
"""
Utility Maximization Problem

----

Problema de Maximização de Utilidade
"""

plt.figure(figsize=(5, 5))
UMP = plt.contour(X, Y, U, levels=[10,20,30],  colors='blue')
B = plt.contour(X, Y, E, levels=[98], colors='black')
plt.clabel(UMP, inline=True, fontsize=8, fmt='%.2f')
plt.clabel(B, inline=True, fontsize=8, fmt='%.2f')
plt.title('Problema de Maximização de Utilidade')
plt.xlabel('x\u2081')
plt.ylabel('x\u2082')
#plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
"""
Expenditure Minimization Problem

----

Problema de Minimização de Gastos
"""

plt.figure(figsize=(5, 5))
EMP = plt.contour(X, Y, E, levels=[50,100,150],  colors='blue')
V = plt.contour(X, Y, U, levels=[20.5], colors='black')
plt.clabel(EMP, inline=True, fontsize=8, fmt='%.2f')
plt.clabel(V, inline=True, fontsize=8, fmt='%.2f')
plt.title('Problema de Minimização de Gastps')
plt.xlabel('x\u2081')
plt.ylabel('x\u2082')
#plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
