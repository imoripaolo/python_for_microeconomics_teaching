```python
"""

Elasticities / Elasticidades

Author / Autor: imoripaolo

"""

import numpy as np
import matplotlib.pyplot as plt

# We'll use a CD function / Vamos usar uma função CD
size = 100
alpha1 = 0.25
alpha2 = 0.75
w = 100

p1 = np.linspace(0.01, 10, size)
p2 = np.linspace(0.01, 15, size)
```

```python
"""

Price Effect / Efeito Preço


"""

x1 = np.zeros(size)
x2 = np.zeros(size)
Dp00 = np.zeros(size)
Dp11 = np.zeros(size)

for i in range(0,size):
    x1[i] = alpha1*w/p1[i]
    x2[i] = alpha2*w/p2[i]
    Dp00[i] = - alpha1*w / p1[i]**2
    Dp11[i] = - alpha2*w / p2[i]**2

fig, ax = plt.subplots() 
ax.plot(p1, Dp00,'b', label='')  
ax.set_ylabel('Dp\u2080\u2080') 
ax.set_xlabel('p\u2081') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
plt.legend(loc='center right')
plt.legend()
plt.show()

fig, ax = plt.subplots() 
ax.plot(x1, Dp00,'b', label='')  
ax.set_ylabel('Dp\u2080\u2080') 
ax.set_xlabel('x\u2081') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
plt.legend(loc='center right')
plt.legend()
plt.show()

fig, ax = plt.subplots() 
ax.plot(p2, Dp11,'b', label='')  
ax.set_ylabel('Dp\u2081\u2081') 
ax.set_xlabel('p\u2082') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
plt.legend(loc='center right')
plt.legend()
plt.show()

fig, ax = plt.subplots() 
ax.plot(x2, Dp11,'b', label='')  
ax.set_ylabel('Dp\u2081\u2081') 
ax.set_xlabel('x\u2082') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
plt.legend(loc='center right')
plt.legend()
plt.show()
```

```python
"""

Price Elasticity of Demand / Elasticidade Preço da Demanda


"""

epsilon11 = np.zeros(size)
epsilon22 = np.zeros(size)

for i in range(0,size):
    epsilon11[i] = Dp11[i]*p1[i]/x1[i]
    epsilon22[i] = Dp22[i]*p2[i]/x2[i]

fig, ax = plt.subplots() 
ax.plot(p1, epsilon11,'b', label='')  
ax.set_ylabel('\u03B5\u2081\u2081') 
ax.set_xlabel('p\u2081') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
plt.legend(loc='center right')
plt.legend()
plt.show()

fig, ax = plt.subplots() 
ax.plot(x1, epsilon11,'b', label='')  
ax.set_ylabel('\u03B5\u2081\u2081') 
ax.set_xlabel('x\u2081') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
plt.legend(loc='center right')
plt.legend()
plt.show()

fig, ax = plt.subplots() 
ax.plot(p2, epsilon22,'b', label='')  
ax.set_ylabel('\u03B5\u2082\u2082') 
ax.set_xlabel('p\u2082') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
plt.legend(loc='center right')
plt.legend()
plt.show()

fig, ax = plt.subplots() 
ax.plot(x2, epsilon22,'b', label='')  
ax.set_ylabel('\u03B5\u2082\u2082') 
ax.set_xlabel('x\u2082') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
plt.legend(loc='center right')
plt.legend()
plt.show()
```
