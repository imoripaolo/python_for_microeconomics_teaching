```python
"""

Elasticities / Elasticidades

Author / Autor: imoripaolo

"""

import numpy as np
import matplotlib.pyplot as plt

# We'll use a CD function / Vamos usar uma função CD
size =100
p1 = 4
p2 = 2
alpha1 = 0.25
alpha2 = 0.75
w = np.linspace(0,100, size)
```

```python
"""

Engel Function / Função de Engel

"""

engel = np.zeros(size)

for i in range(0,size):
    engel[i] = alpha1*w[i]/p1

fig, ax = plt.subplots() 
ax.plot(engel,w,'b') 
ax.set_ylabel('w') 
ax.set_xlabel('x\u2081') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
plt.show()
```

```python
"""

Income Expansion Path / Caminho de Expansão da Renda

"""

iep1 = np.zeros(size)
iep2 = np.zeros(size)

for i in range(0,size):
    iep1[i] = alpha1*w[i]/p1
    iep2[i] = alpha2*w[i]/p2

fig, ax = plt.subplots() 
ax.plot(iep1,iep2,'b') 
ax.set_ylabel('x\u2082') 
ax.set_xlabel('x\u2081') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
plt.show()
```

```python
"""

Income Effect / Efeito Renda

"""

Dw = ( alpha1 / p1, alpha2 / p2)
print(Dw)
```

```python
"""

Income-Elasticity of Demand / Elasticidade-Renda da Demanda

Aditional Example: Quasilinear Utility / Exemplo Adicional: Utilidade Quasilinear
u(x1,x2) = ln (x1) + x2

"""

x11 = np.zeros(size)
x21 = np.zeros(size)
eta11 = np.zeros(size)
eta21 = np.zeros(size)

for i in range(0,size):
    x11[i] = alpha1*w[i]/p1
    x21[i] = alpha2*w[i]/p2
    eta11[i] = Dw[0]*w[i]/x11[i]
    eta11[i] = Dw[1]*w[i]/x21[i]

fig, ax = plt.subplots() 
ax.plot(w,eta11,'b', label='\u03B7\u2081') 
ax.plot(w,eta21,'r', label='\u03B7\u2082') 
ax.set_ylabel('\u03B7\u2081, \u03B7\u2082') 
ax.set_xlabel('w') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
plt.legend(loc='center right')
plt.legend()
plt.show()

x12 = np.zeros(size)
x22 = np.zeros(size)
eta12 = np.zeros(size)
eta22 = np.zeros(size)
Dw2 = ( 1 / p1 , 0 )

for i in range(0,size):
    x12[i] = ( w[i] - p1 ) / p1
    x22[i] = p1 / p2
    eta12[i] = Dw2[0]*w[i]/x12[i]
    eta22[i] = Dw2[1]*w[i]/x22[i]

fig, ax = plt.subplots() 
ax.plot(w,eta12,'b', label='\u03B7\u2081')  
ax.set_ylabel('\u03B7\u2081, \u03B7\u2082') 
ax.set_xlabel('w') 
plt.axhline(y=0, color='black') 
plt.axvline(x=0, color='black') 
ax.plot(w,eta22,'r', label='\u03B7\u2082')
plt.legend(loc='center right')
plt.legend()
plt.show()
```
